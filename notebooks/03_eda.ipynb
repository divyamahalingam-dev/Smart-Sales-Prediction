# 03. Exploratory Data Analysis (EDA)
**Smart Sales Prediction and Demand Forecasting System**

Visualizing trends, seasonality, store performance, holiday surges, and promotional lifts.


```python
import sys, os
sys.path.append('../src')
import pandas as pd
import plotly.express as px

clean_df = pd.read_csv('../data/raw_sales.csv')
clean_df['date'] = pd.to_datetime(clean_df['date'])
clean_df.info()
```

```python
# Monthly Revenue Trend
clean_df['year_month'] = clean_df['date'].dt.to_period('M').dt.to_timestamp()
monthly = clean_df.groupby('year_month')['revenue'].sum().reset_index()
fig = px.line(monthly, x='year_month', y='revenue', title='Monthly Revenue Trend')
fig.show()
```

```python
# Day of Week Demand Distribution
dow_names = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
clean_df['dow'] = clean_df['date'].dt.dayofweek.map(lambda x: dow_names[x])
dow_summary = clean_df.groupby('dow')['quantity'].mean().reindex(dow_names).reset_index()
fig_dow = px.bar(dow_summary, x='dow', y='quantity', title='Average Sales Quantity by Day of Week')
fig_dow.show()
```

```python
# Promotional Lift
promo_comp = clean_df.groupby('promotion')['quantity'].agg(['mean', 'median', 'std'])
print('Promotional Uplift Table:')
print(promo_comp)
```
