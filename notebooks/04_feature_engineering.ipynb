# 04. Feature Engineering for Demand Forecasting
**Smart Sales Prediction and Demand Forecasting System**

Constructing temporal calendar features, lag features (1, 7, 14, 30 days), rolling statistics, and interaction terms.


```python
import sys, os
sys.path.append('../src')
import pandas as pd
from feature_engineering import FeatureEngineer

raw_df = pd.read_csv('../data/raw_sales.csv')
raw_df['date'] = pd.to_datetime(raw_df['date'])

fe = FeatureEngineer()
X, y, feature_cols = fe.fit_transform(raw_df)
print(f'Constructed {len(feature_cols)} features across {len(X):,} records.')
print('Feature List:', feature_cols)
```

```python
# Inspect correlation of top features with sales quantity
corr_df = X.copy()
corr_df['target_quantity'] = y
corrs = corr_df.corr()['target_quantity'].sort_values(ascending=False)
print('Top Positive & Negative Correlations with Demand:')
print(corrs.head(10))
print(corrs.tail(5))
```
