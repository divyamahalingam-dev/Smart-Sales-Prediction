This notebook handles:
1. Deduplication and data integrity verification.
2. Missing value handling and domain constraint validations.
3. Outlier detection and capping.
4. Mathematical consistency of price, discount, quantity, and revenue.


```python
import sys, os
sys.path.append('../src')
import pandas as pd
from preprocessing import clean_sales_data

raw_path = '../data/raw_sales.csv'
raw_df = pd.read_csv(raw_path)
print(f'Raw dataset shape: {raw_df.shape}')
raw_df.head()
```

```python
# Run preprocessing
cleaned_df, stats = clean_sales_data(raw_df, cap_outliers=True)
print('Cleaning Summary:')
for k, v in stats.items():
    print(f'  {k}: {v}')
```

```python
# Verify numerical constraints and revenue formula
revenue_check = (cleaned_df['quantity'] * cleaned_df['unit_price'] * (1.0 - cleaned_df['discount'])).round(2)
diff = (cleaned_df['revenue'] - revenue_check).abs().max()
print(f'Maximum discrepancy in revenue calculation: {diff:.4f}')
```
